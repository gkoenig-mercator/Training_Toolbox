## Using the describe command

The **describe** command retrieves the metadata of the data available on Copernicus Marine.

You can browse the catalogue of all products on the [Data Store website](https://data.marine.copernicus.eu/products). This is where you will find the product_id or dataset_id you need to retrieve the metadata of a given product or dataset.

### General use

The **describe** command returns a JSON object, which can contain a lot of information. JSON is practical because the information is organised hierarchically, by section. Without any argument, the command returns the metadata of the whole catalogue: the output is very large and **it will noticeably slow down the notebook**, so we advise you not to run the cell below.

```python
!copernicusmarine describe
```

This output is hard to read as it is. In the video below, we explore it in a web browser instead, which is easier. To do so, you first need to save the output to a file, which is easiest from the command line (CLI). The `>` operator redirects the output of a command into a file instead of printing it on screen.

```python
!copernicusmarine describe > catalogue.json
```

## Product information

You can also restrict the output to a single product. As this training focuses on polar regions, we will use the [Arctic Ocean sea ice analysis and forecast product](https://data.marine.copernicus.eu/product/ARCTIC_ANALYSISFORECAST_PHY_ICE_002_011/services).

```python
metadata_arctic= copernicusmarine.describe(product_id="ARCTIC_ANALYSISFORECAST_PHY_ICE_002_011")
```

Let's look at the raw output of the command. To read it more easily, right-click on the output and select **Enable Scrolling for Outputs**.

```python
metadata_arctic
```

The output is organised hierarchically: a list of products, each containing information about the product and a list of datasets, each containing information about the dataset (and, further down, its versions, parts, services and variables). For example, you can get the short description of the product with the following command:

```python
metadata_arctic.products[0].description
```

You can access other fields in the same way, such as the processing level or the link to the product user manual (PUM):

```python
metadata_arctic.products[0].processing_level
```

```python
metadata_arctic.products[0].product_user_manual
```

## Dataset information

You can also request information about a single dataset. In that case, the returned object has the same structure as before: a list of products containing only one product, which itself contains only your dataset. Here is an example with an in-situ dataset covering the whole globe:

```python
insitu_metadata = copernicusmarine.describe(dataset_id="cmems_obs-ins_glo_phy-cur_my_adcp_irr")
```

We can now access the metadata of this dataset:

```python
insitu_metadata.products[0].datasets[0]
```

If we want information about the variables, for example, we can loop over the variables of the dataset:

```python
list_of_variables = []
for var in insitu_metadata.products[0].datasets[0].versions[0].parts[0].services[0].variables:
    list_of_variables.append(var.standard_name)

print(list_of_variables)
```

### Exercise

Get the list of bounding boxes (`bbox`) associated with the variables listed above.

### Solution

```python
list_of_bboxes = []
for var in insitu_metadata.products[0].datasets[0].versions[0].parts[0].services[0].variables:
    list_of_bboxes.append(var.bbox)

print(list_of_bboxes)
```

## Searching the catalogue

So far, we already knew which product or dataset we wanted. When you don't, the **contains** option of describe lets you search the whole catalogue: it keeps only the products whose metadata contain the string you give. Note that the command still has to go through the whole catalogue, so it can take a minute.

### Exercise

Find all the **Arctic** products whose metadata mention **salinity**. For each of them, print its product_id, its title and its number of datasets.

<details>
<summary><b>Hint</b> (click to expand)</summary>

Search with `contains=["salinity"]`, then loop over the products and keep those whose title contains "Arctic". The number of datasets is the length of the `datasets` list of each product.
</details>

### Solution

```python
salinity_catalogue = copernicusmarine.describe(contains=["salinity"])

for product in salinity_catalogue.products:
    if "arctic" in product.title.lower():
        print(f"{product.product_id} - {product.title} ({len(product.datasets)} datasets)")
```
