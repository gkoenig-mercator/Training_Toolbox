## Use of the get command

The get command is used to retrieve data from the data lake of Copernicus Marine. The data retrieved by get are different from those retrieved by the subset as they are the **original data** as delivered by the data producers

You can find the catalogue of all data at the [website of the Data Store](https://data.marine.copernicus.eu/products?disc=facets). There, you will need the **dataset_id** as well as some information about the name format to actually retrieve a file.  

### Getting information about a dataset

We will try to retrieve physical modelization of the Antarctic ocean and gather information about those using the get command. In the following video we will show how to find a dataset and find the structure of the data using the link provided above. 

```python
Video("videos/Looking_for_data_info.webm", width=1200, html_attributes="controls")
```

Now that we have identified the dataset,we can use the **dry_run** option that gives the same output as the command, but without actually performing the downloading. It allows us to check the number of files and their sizes for example.

```python
get_output_example = copernicusmarine.get(dataset_id="cmems_mod_arc_phy_anfc_6km_detided_P1D-m", dry_run=True)
```

And if we inspect the output. It would be nice to right-click on the cell and select "Enable Scrolling for Outputs".

```python
get_output_example
```

As you can see, we see a long list of files, with informations about all of them. At the end are informations about the number of files and the total sizes of files to download. Let's inspect it.

```python
print(f"The total number of files is {get_output_example.number_of_files_to_download} and the total size of download is {int(get_output_example.total_size/1000)} GB.")
```

We can also inspect information about just a file. Their information are in the list of the ResponseGet object and we can see what are the file names and their sizes

```python
print(f"The file called '{get_output_example.files[12].filename}' has a size of {int(get_output_example.files[12].file_size)} MB")
```

### Selecting files

We see that downloading the entire dataset is out of reach for a teaching session. We should focus on just downloading one or two files.

For this, we can use the **filter** subcommand. It allows you to select files depending on a pattern found in their name. We will try to select the file "20210717_dm-metno-MODEL-topaz5-ARC-b20210719-fv02.0.nc" mentioned above with this. 

After a quick inspection of the other files, we see that most of the name if common but that there is one file per day for this dataset. However, there are two dates in the name: They can correspond to the start time and the end time of the data. It means that if we just select a date, there is likely another file whose name will also contain that date. Here, we can decide to choose the first date with the "_" as our pattern, as the second date is surrounded by "-". For the functioning of the filtering option, you have to surround your pattern with "*" which means that there could any string of characters before and after your pattern. For more explanation about this, you can look at [the "Standard Wildcard" section of this website](https://tldp.org/LDP/GNU-Linux-Tools-Summary/html/x11655.htm).

```python
get_output_example_filter = copernicusmarine.get(dataset_id="cmems_mod_arc_phy_anfc_6km_detided_P1D-m", filter = "*20210717_*",dry_run=True)
```

And we can inspect the ouput

```python
get_output_example_filter
```

And see that we have one file of 242 MB

```python
print(f"The total number of files is {get_output_example_filter.number_of_files_to_download} and the total size of download is {int(get_output_example_filter.total_size)} MB.")
```

### Downloading and inspecting files

Now we can actually download and inspect one of the files. There are different types of files on the Copernicus Marine Data Store. The data we are interested in here are models outputs that are stored in the NetCDF format. You can find more information about this format on [its official website](https://www.unidata.ucar.edu/software/netcdf). For reading those data, we will need to use a package that is out of the toolbox, here [xarray in Python](https://xarray.dev/). 

First we actually download the file by removing the **dry_run** option and not storing the output in a variable, as we will not need it

```python
copernicusmarine.get(dataset_id="cmems_mod_arc_phy_anfc_6km_detided_P1D-m", filter = "*20210717_*")
```

Now we can open the file with xarray to see what it contains. You can see that the file is in a succession of directories, as the get command copies the directory organization of the datalake. As a small tip, the path is indicated is the field **file_path**

```python
physical_data = xr.open_dataset("ARCTIC_ANALYSISFORECAST_PHY_002_001/cmems_mod_arc_phy_anfc_6km_detided_P1D-m_202311/2021/07/20210717_dm-metno-MODEL-topaz5-ARC-b20210719-fv02.0.nc")
```

```python
physical_data
```

There are other datasets but they only contain data for the last months or few weeks before beind included in other datasets. We will see what is in the dataset with the --dry-run option

```python
example_satellite_plankton = copernicusmarine.get(dataset_id="cmems_obs-oc_arc_bgc-plankton_my_l4-multi-4km_P1M",dry_run=True)
```

```python
example_satellite_plankton
```

```python
print(f"The total number of files is {example_satellite_plankton.number_of_files_to_download} and the total size of download is {int(example_satellite_plankton.total_size)} MB.")
```

We see in the file names that the files correspond to a month of data, with date patterns such as "20240401-20240430". We will choose the month of July 2024, it should contain the pattern **20240701-20240731**.

```python
copernicusmarine.get(dataset_id="cmems_obs-oc_arc_bgc-plankton_my_l4-multi-4km_P1M",filter="*20240701-20240731*")
```

And now we can open the file to inspect it

```python
plankton_satellite_data = xr.open_dataset('OCEANCOLOUR_ARC_BGC_L4_MY_009_124/cmems_obs-oc_arc_bgc-plankton_my_l4-multi-4km_P1M_202311/2024/20240701-20240731_cmems_obs-oc_arc_bgc-plankton_my_l4-multi-4km_P1M.nc')
```
